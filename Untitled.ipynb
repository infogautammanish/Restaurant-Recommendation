```python
# Cell 6: Data Cleaning - Fix currency column
# The currency column has incorrect values for Philippines
# Let's create a function to clean this
def clean_currency(row):
    if row['Country Code'] == 162:  # Philippines
        return 'Philippine Peso(PHP)'
    elif row['Country Code'] == 30:  # Brazil
        return 'Brazilian Real(R$)'
    elif row['Country Code'] == 216:  # USA
        return 'Dollar($)'
    elif row['Country Code'] == 14:  # Australia
        return 'Dollar($)'
    elif row['Country Code'] == 1:  # India
        return 'Indian Rupees(Rs.)'
    elif row['Country Code'] == 214:  # UAE
        return 'Emirati Diram(AED)'
    elif row['Country Code'] == 184:  # Singapore
        return 'Dollar($)'
    elif row['Country Code'] == 37:  # Canada
        return 'Dollar($)'
    else:
        return row['Currency']

df['Currency'] = df.apply(clean_currency, axis=1)
print("Currency column cleaned!")
df['Currency'].value_counts()
```

**TASK 2: Restaurant Recommendation System**

```python
# Cell 14: Task 2 - Prepare data for recommendation system
df_task2 = df.copy()

# Clean cuisines column
def clean_cuisine(cuisine_str):
    if pd.isna(cuisine_str) or cuisine_str == 'Unknown':
        return []
    # Split by comma and strip whitespace
    cuisines = [c.strip() for c in cuisine_str.split(',')]
    return cuisines

df_task2['Cuisines_List'] = df_task2['Cuisines'].apply(clean_cuisine)

# Create a binary matrix for cuisines
mlb = MultiLabelBinarizer()
cuisine_matrix = mlb.fit_transform(df_task2['Cuisines_List'])
cuisine_df = pd.DataFrame(cuisine_matrix, columns=mlb.classes_, index=df_task2.index)

# Combine with other features
df_task2_features = pd.concat([
    cuisine_df,
    pd.get_dummies(df_task2['City'], prefix='City'),
    pd.get_dummies(df_task2['Price range'], prefix='Price')
], axis=1)

print("Features for recommendation system:")
print(f"Shape: {df_task2_features.shape}")
```

```python
# Cell 15: Task 2 - Define recommendation function
def recommend_restaurants(preferred_cuisines, preferred_city=None, preferred_price=None, 
                          top_n=10, features_df=df_task2_features, original_df=df_task2):
    """
    Recommend restaurants based on user preferences using content-based filtering.
    
    Parameters:
    - preferred_cuisines: list of cuisine names (e.g., ['Italian', 'Mexican'])
    - preferred_city: city name (e.g., 'New Delhi')
    - preferred_price: price range (1-4)
    - top_n: number of recommendations to return
    - features_df: feature matrix
    - original_df: original dataframe with restaurant info
    
    Returns:
    - DataFrame with top N restaurant recommendations
    """
    
    # Create user preference vector
    user_vector = np.zeros(features_df.shape[1])
    
    # Add cuisine preferences
    cuisine_cols = mlb.classes_
    for cuisine in preferred_cuisines:
        if cuisine in cuisine_cols:
            user_vector[features_df.columns.get_loc(cuisine)] = 1
    
    # Add city preference
    if preferred_city:
        city_col = f'City_{preferred_city}'
        if city_col in features_df.columns:
            user_vector[features_df.columns.get_loc(city_col)] = 1
    
    # Add price preference
    if preferred_price:
        price_col = f'Price_{preferred_price}'
        if price_col in features_df.columns:
            user_vector[features_df.columns.get_loc(price_col)] = 1
    
    # Calculate cosine similarity
    similarities = cosine_similarity([user_vector], features_df)[0]
    
    # Get top N most similar restaurants
    top_indices = similarities.argsort()[-top_n:][::-1]
    
    # Return recommendations with details
    recommendations = original_df.iloc[top_indices][
        ['Restaurant Name', 'City', 'Cuisines', 'Average Cost for two', 
         'Aggregate rating', 'Votes', 'Price range']
    ].copy()
    recommendations['Similarity Score'] = similarities[top_indices]
    
    return recommendations
```

```python
# Cell 16: Task 2 - Test the recommendation system
# Example 1: Italian food in New Delhi with moderate price
print("=" * 80)
print("Recommendation Example 1: Italian cuisine in New Delhi, Price range 3")
print("=" * 80)
recommendations1 = recommend_restaurants(
    preferred_cuisines=['Italian', 'Pizza'],
    preferred_city='New Delhi',
    preferred_price=3,
    top_n=5
)
print(recommendations1)

print("\n" + "=" * 80)
print("Recommendation Example 2: Japanese food in Dubai, any price")
print("=" * 80)
recommendations2 = recommend_restaurants(
    preferred_cuisines=['Japanese', 'Sushi'],
    preferred_city='Dubai',
    top_n=5
)
print(recommendations2)

print("\n" + "=" * 80)
print("Recommendation Example 3: Mexican food in any city, affordable price")
print("=" * 80)
recommendations3 = recommend_restaurants(
    preferred_cuisines=['Mexican'],
    preferred_price=1,
    top_n=5
)
print(recommendations3)
```

**TASK 3: Cuisine Classification**

```python
# Cell 17: Task 3 - Prepare data for cuisine classification
df_task3 = df.copy()

# Create a copy for classification
df_task3_class = df_task3.copy()

# Only keep rows where cuisine is known
df_task3_class = df_task3_class[df_task3_class['Cuisines'] != 'Unknown']

# Extract primary cuisine (first cuisine listed)
df_task3_class['Primary_Cuisine'] = df_task3_class['Cuisines'].apply(
    lambda x: str(x).split(',')[0].strip() if pd.notna(x) else 'Unknown'
)

# Keep only top 10 cuisines for better classification
top_cuisines = df_task3_class['Primary_Cuisine'].value_counts().head(10).index
df_task3_class = df_task3_class[df_task3_class['Primary_Cuisine'].isin(top_cuisines)]

print(f"Dataset for cuisine classification: {df_task3_class.shape}")
print(f"Target cuisines: {top_cuisines.tolist()}")
```

```python
# Cell 18: Task 3 - Feature engineering for cuisine classification
# Create feature set
features_task3 = df_task3_class.copy()

# Drop unnecessary columns
columns_to_drop = ['Restaurant ID', 'Restaurant Name', 'Address', 'Locality', 
                   'Locality Verbose', 'Cuisines', 'Rating color', 'Rating text',
                   'Primary_Cuisine']  # Keep this as target
features_task3 = features_task3.drop(columns=columns_to_drop)

# Encode categorical variables
categorical_cols = features_task3.select_dtypes(include=['object']).columns
features_task3_encoded = pd.get_dummies(features_task3, columns=['City', 'Currency', 
                                                                  'Has Table booking',
                                                                  'Has Online delivery',
                                                                  'Is delivering now',
                                                                  'Switch to order menu'],
                                        drop_first=True)

# Label encode Country Code
le_country = LabelEncoder()
features_task3_encoded['Country Code'] = le_country.fit_transform(features_task3_encoded['Country Code'])

print(f"Feature shape: {features_task3_encoded.shape}")
```

```python
# Cell 19: Task 3 - Split data and scale features
X = features_task3_encoded
y = df_task3_class['Primary_Cuisine']

# Label encode target
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Scale numerical features
scaler_task3 = StandardScaler()
numerical_cols_task3 = ['Average Cost for two', 'Votes', 'Price range']
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numerical_cols_task3] = scaler_task3.fit_transform(X_train[numerical_cols_task3])
X_test_scaled[numerical_cols_task3] = scaler_task3.transform(X_test[numerical_cols_task3])

print(f"Training set: {X_train.shape}")
print(f"Testing set: {X_test.shape}")
```

```python
# Cell 20: Task 3 - Train classification models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

classifiers = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000, multi_class='multinomial'),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100, max_depth=10)
}

results_task3 = {}

for name, clf in classifiers.items():
    # Train
    clf.fit(X_train_scaled, y_train)
    
    # Predict
    y_pred = clf.predict(X_test_scaled)
    
    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')
    
    results_task3[name] = {
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1
    }
    
    print(f"\n{name} Results:")
    print(f"  Accuracy: {accuracy:.4f}")
    print(f"  Precision: {precision:.4f}")
    print(f"  Recall: {recall:.4f}")
    print(f"  F1-Score: {f1:.4f}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, target_names=le_target.classes_))
```

```python
# Cell 21: Task 3 - Analyze model performance by cuisine
# Get predictions from Random Forest
rf_clf = classifiers['Random Forest']
y_pred_rf = rf_clf.predict(X_test_scaled)

# Create confusion matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred_rf)

# Plot confusion matrix
plt.figure(figsize=(12, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=le_target.classes_,
            yticklabels=le_target.classes_)
plt.title('Confusion Matrix - Cuisine Classification')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Show per-class performance
print("\nPer-Cuisine Performance:")
for i, cuisine in enumerate(le_target.classes_):
    precision = cm[i, i] / cm[:, i].sum() if cm[:, i].sum() > 0 else 0
    recall = cm[i, i] / cm[i, :].sum() if cm[i, :].sum() > 0 else 0
    print(f"{cuisine:20s} - Precision: {precision:.3f}, Recall: {recall:.3f}")
```
